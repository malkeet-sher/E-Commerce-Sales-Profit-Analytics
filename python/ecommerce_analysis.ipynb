```python
import numpy as np

# Start again from the untouched raw dataset
df_clean = df.copy()

# 1. Standardize text columns
df_clean["Category"] = (
    df_clean["Category"]
    .astype("string")
    .str.strip()
    .str.title()
)

df_clean["Region"] = (
    df_clean["Region"]
    .astype("string")
    .str.strip()
    .str.title()
)

df_clean["Channel"] = (
    df_clean["Channel"]
    .astype("string")
    .str.strip()
    .str.title()
)

# 2. Recover missing Customer_ID
df_clean["Customer_ID"] = df_clean["Customer_ID"].fillna("UNKNOWN")

# 3. Recover missing Category using the product's most common category
category_map = (
    df_clean.dropna(subset=["Category"])
    .groupby("Product")["Category"]
    .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else "UNKNOWN")
)

df_clean["Category"] = (
    df_clean["Category"]
    .fillna(df_clean["Product"].map(category_map))
    .fillna("UNKNOWN")
)

# 4. Missing Region cannot be reliably inferred from Product
df_clean["Region"] = df_clean["Region"].fillna("UNKNOWN")

# 5. Correctly convert Discount
#    0.05 stays 0.05
#    "5%" becomes 0.05
discount_raw = df_clean["Discount"].astype("string").str.strip()
is_percent = discount_raw.str.endswith("%", na=False)

df_clean["Discount"] = pd.to_numeric(
    discount_raw.str.replace("%", "", regex=False),
    errors="coerce"
)

df_clean.loc[is_percent, "Discount"] = (
    df_clean.loc[is_percent, "Discount"] / 100
)

# 6. Recover missing Discount from Sales where possible
discount_from_sales = (
    1 - df_clean["Sales"] /
    (df_clean["Quantity"] * df_clean["Unit_Price"])
)

df_clean["Discount"] = df_clean["Discount"].fillna(discount_from_sales)

# 7. Convert dates
df_clean["Order_Date"] = pd.to_datetime(
    df_clean["Order_Date"],
    format="mixed",
    errors="coerce"
)

# 8. Recover missing Quantity
quantity_from_sales = (
    df_clean["Sales"] /
    (df_clean["Unit_Price"] * (1 - df_clean["Discount"]))
)

df_clean["Quantity"] = df_clean["Quantity"].fillna(quantity_from_sales)

# 9. Recover missing Sales
sales_from_formula = (
    df_clean["Quantity"] *
    df_clean["Unit_Price"] *
    (1 - df_clean["Discount"])
)

df_clean["Sales"] = df_clean["Sales"].fillna(sales_from_formula)

# 10. Remove exact duplicate records
df_clean = df_clean.drop_duplicates().copy()

# 11. Identify sales and returns
df_clean["Order_Type"] = np.where(
    df_clean["Quantity"] < 0,
    "Return",
    "Sale"
)

# 12. Calculate Profit and Profit Margin
df_clean["Profit"] = df_clean["Sales"] - df_clean["Cost"]

df_clean["Profit_Margin"] = np.where(
    df_clean["Order_Type"] == "Sale",
    df_clean["Profit"] / df_clean["Sales"],
    np.nan
)

# Recreate analysis subsets
sales_only = df_clean[df_clean["Order_Type"] == "Sale"].copy()
returns = df_clean[df_clean["Order_Type"] == "Return"].copy()

# Save corrected clean dataset
df_clean.to_csv(
    "../data/ecommerce_orders_clean.csv",
    index=False
)

print("CORRECTED CLEAN DATA")
print("--------------------")
print("Rows:", len(df_clean))
print("Columns:", len(df_clean.columns))
print("Missing values:", df_clean.isnull().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
print("Sales orders:", len(sales_only))
print("Return orders:", len(returns))

print("\nDiscount values:")
print(df_clean["Discount"].value_counts().sort_index())
```

## Business Findings

### Overall Performance
- Net sales and net profit were calculated after cleaning the transaction data.
- Returns were identified separately from normal sales.
- Overall profit margin was also calculated for sale transactions.

### Product Performance
- Laptop generated the highest sales and profit contribution.
- Mouse had a higher profit margin despite much lower revenue.
- This shows that sales value and profitability should be analyzed together.

### Category Performance
- Electronics generated substantially more revenue and profit than Accessories.
- Accessories had high unit volume but lower revenue because of lower-priced products.

### Monthly Performance
- Monthly sales and profit were compared to identify stronger and weaker periods.
- The monthly analysis helps identify changes in business performance over time.

### Channel Performance
- App and Website generated similar sales and profit.
- Marketplace contributed less sales and profit than the other two channels.

### Return Analysis
- 71 orders were identified as returns.
- Returns created a negative impact on sales and profit.
- Return activity should therefore be monitored separately from normal sales.

```python
query = """
SELECT
    COUNT(*) AS Repeat_Customers
FROM (
    SELECT
        Customer_ID
    FROM orders
    WHERE Order_Type = 'Sale'
      AND Customer_ID <> 'UNKNOWN'
    GROUP BY Customer_ID
    HAVING COUNT(*) > 1
);
"""

pd.read_sql_query(query, conn)
```

```python
query = """
WITH product_profit AS (
    SELECT
        Category,
        Product,
        SUM(Sales) AS Total_Sales,
        SUM(Profit) AS Total_Profit
    FROM orders
    WHERE Order_Type = 'Sale'
    GROUP BY Category, Product
)

SELECT
    Category,
    Product,
    ROUND(Total_Sales, 2) AS Total_Sales,
    ROUND(Total_Profit, 2) AS Total_Profit,
    RANK() OVER (
        PARTITION BY Category
        ORDER BY Total_Profit DESC
    ) AS Profit_Rank
FROM product_profit
ORDER BY Category, Profit_Rank;
"""

pd.read_sql_query(query, conn)
```
